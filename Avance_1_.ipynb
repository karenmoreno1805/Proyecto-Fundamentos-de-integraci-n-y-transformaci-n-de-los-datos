```python
# ------------------------------------------------------------------------------
# CELDA 3: Componente 01 - Definición del Problema y Objetivos del Pipeline
# ------------------------------------------------------------------------------
print("""
================================================================================
COMPONENTES DEL AVANCE 1: MARCO DEL PROBLEMA Y OBJETIVOS
================================================================================
PROBLEMA IDENTIFICADO:
Existe una desconexión estructural entre la información de buscadores de empleo y las
vacantes demandadas en el Tolima. La información está distribuida en fuentes heterogéneas
(macrodatos del DANE y microdatos del Servicio Público de Empleo / ComfaTolima).

OBJETIVO GENERAL DEL PIPELINE:
Diseñar y construir un pipeline de integración y transformación centralizado para
consolidar datos macroeconómicos y operativos del mercado laboral en el Tolima.
================================================================================
""")
```

```python
# ------------------------------------------------------------------------------
# CELDA 4: Componente 02 & 03 - Extracción Inicial Técnica de los Orígenes
# ------------------------------------------------------------------------------
print("=== INICIANDO EXTRACTION TÉCNICA DE FUENTES ===")
```

```python
# 1. Carga de ComfaTolima
df_comfa_raw = pd.read_excel(path_comfatolima)
print(f"[OK] COMFATOLIMA cargado exitosamente.")
print(f"     Dimensiones: {df_comfa_raw.shape[0]} filas x {df_comfa_raw.shape[1]} columnas.")
```

```python
# 2. Carga del DANE
df_dane_raw = pd.read_excel(path_dane, sheet_name='Hoja1')
print(f"[OK] DANE cargado exitosamente.")
print(f"     Dimensiones: {df_dane_raw.shape[0]} filas x {df_dane_raw.shape[1]} columnas.\n")
```

```python
# ------------------------------------------------------------------------------
# CELDA 5: Componente 04 - Construcción Automatizada del Diccionario de Datos
# ------------------------------------------------------------------------------
print("=== DICCIONARIO DE DATOS AUTO-GENERADO: COMFATOLIMA ===")
dict_comfa = []
for col in df_comfa_raw.columns:
    dict_comfa.append({
        'Variable': col,
        'Tipo_Dato_Original': str(df_comfa_raw[col].dtype),
        'Nulos': df_comfa_raw[col].isnull().sum(),
        'Valores_Unicos': df_comfa_raw[col].nunique(),
        'Ejemplo_Valor': str(df_comfa_raw[col].iloc[0])[:40]
    })
df_dict_comfa = pd.DataFrame(dict_comfa)
display(df_dict_comfa)

print("\n=== DICCIONARIO DE DATOS AUTO-GENERADO: DANE ===")
dict_dane = []
for col in df_dane_raw.columns:
    dict_dane.append({
        'Columna': col,
        'Tipo_Dato_Original': str(df_dane_raw[col].dtype),
        'Nulos': df_dane_raw[col].isnull().sum(),
        'Ejemplo_Valor': str(df_dane_raw[col].iloc[2])[:40]
    })
df_dict_dane = pd.DataFrame(dict_dane)
display(df_dict_dane.head(5))
```

```python
# ------------------------------------------------------------------------------
# CELDA 6: Componentes 04 y 05 - Diagnóstico de Calidad y Evidencia Crítica
# ------------------------------------------------------------------------------
print("=== DIAGNÓSTICO AUTOMATIZADO DE CALIDAD DE DATOS ===")
```

```python
# 1. Duplicados
dup_filas = df_comfa_raw.duplicated().sum()
dup_vacantes = df_comfa_raw.duplicated(subset=['CODIGO_VACANTE']).sum()
print(f"[EVIDENCIA 1] Filas totalmente duplicadas en ComfaTolima: {dup_filas}")
print(f"[EVIDENCIA 1] Códigos de vacante duplicados: {dup_vacantes}")
```

```python
# 2. Registros Geográficos (Tolima vs Nacional)
ofertas_tolima = df_comfa_raw[df_comfa_raw['DEPARTAMENTO'].str.upper() == 'TOLIMA']
vacantes_totales_tolima = ofertas_tolima['CANTIDAD_VACANTES'].sum()
print(f"[EVIDENCIA 2] Ofertas con departamento 'TOLIMA': {len(ofertas_tolima)} de {len(df_comfa_raw)} ({len(ofertas_tolima)/len(df_comfa_raw)*100:.1f}%)")
print(f"[EVIDENCIA 2] Total de puestos/plazas de trabajo en Tolima: {vacantes_totales_tolima}")
```

```python
# 3. Outliers en Experiencia Requerida
outliers_exp = df_comfa_raw[df_comfa_raw['MESES_EXPERIENCIA_CARGO'] > 120]
print(f"[EVIDENCIA 3] Vacantes con experiencia > 120 meses (10 años): {len(outliers_exp)}")
print(f"              Valor máximo registrado: {df_comfa_raw['MESES_EXPERIENCIA_CARGO'].max()} meses")
```

```python
# 4. Ambigüedad en Salarios
salarios_convenir = (df_comfa_raw['RANGO_SALARIAL'] == 'A Convenir').sum()
print(f"[EVIDENCIA 4] Vacantes con salario 'A Convenir': {salarios_convenir} ({salarios_convenir/len(df_comfa_raw)*100:.1f}%)")
```
